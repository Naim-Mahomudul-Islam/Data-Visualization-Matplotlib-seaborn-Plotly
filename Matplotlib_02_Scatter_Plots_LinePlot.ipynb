Use  to find correlation between features:

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
students = pd.read_csv('student_IQdata.csv')
students
```

```python
plt.scatter(students['Study_Hour'], students['IQ_Score'])
```

```python

#Relation between study Hours and IQ score


#Positive relation Exesist  :::::::::::        +++++++++++++++++++++   ::::::::::

plt.xlabel('Study_Hour')
plt.ylabel('IQ_Score')
plt.title('Relation between Study_Hour and IQ score ')





plt.scatter(students['Study_Hour'], students['IQ_Score'])
```

```python
#Relation between Shoe_Size and IQ score


#No relation Exesist

plt.xlabel('Shoe_Size')
plt.ylabel('IQ_Score')
plt.title('Relation between Shoe_Size and IQ score ')





plt.scatter(students['Shoe_Size'], students['IQ_Score'])
```

```python
#Relation between Chilling_Hours  and IQ score

#Negative Relation : 

plt.xlabel('Chilling_Hours')
plt.ylabel('IQ_Score')
plt.title('Relation between Chilling_Hours and IQ score ')





plt.scatter(students['Chilling_Hours'], students['IQ_Score'])
```

```python
#Relation between Chilling_Hours  and IQ score

#Negative Relation : 

plt.xlabel('Chilling_Hours')
plt.ylabel('IQ_Score')
plt.title('Relation between Chilling_Hours and IQ score ')





plt.scatter(students['Chilling_Hours'], students['IQ_Score'],color='red')
```
