```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
students = pd.read_csv('student_IQdata.csv')
students
```

```python
plt.xlabel('IQ_Score')
plt.ylabel('Frequency of Students')
plt.title('Frequency of Students base on IQ_Score ')


plt.hist(students['IQ_Score'])

#140 outlear 
```

```python
plt.xlabel('IQ_Score')
plt.ylabel('Frequency of Students')
plt.title('Frequency of Students base on IQ_Score ')


plt.hist(students['IQ_Score'],color='green',edgecolor='black')

#140 outlear 
```

```python
plt.xlabel('Study_Hour')
plt.ylabel('Frequency of Students')
plt.title('Frequency of Students base on IQ_Score ')


plt.hist(students['Study_Hour'],color='green',edgecolor='black')

#140 outlear 
```

```python
#select bin size :


plt.xlabel('Study_Hour')
plt.ylabel('Frequency of Students')
plt.title('Frequency of Students base on IQ_Score ')


plt.hist(students['Study_Hour'],bins=5,color='green',edgecolor='black')

#140 outlear 




```

```python
#select bin size :


plt.xlabel('Study_Hour')
plt.ylabel('Frequency of Students')
plt.title('Frequency of Students base on IQ_Score ')


plt.hist(students['Study_Hour'],bins=20,color='green',edgecolor='black')

#140 outlear 




```
